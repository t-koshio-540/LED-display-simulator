```python
from PIL import Image
import numpy as np
import tkinter as tk
```

```python
root = tk.Tk()
root.geometry("1440x320")

canvas = tk.Canvas(root, bg="gray")
canvas.pack(fill = tk.BOTH, expand = True)

canvas.create_rectangle((0,0),(1441,321),fill="black")

def WriteAll():
    img = Image.open('./image/EDN2000VR20509.bmp')
    palette=img.getpalette()
    palette=np.array(palette).reshape(-1,3)
    width=img.size[0]
    height=img.size[1]
    for x in range(width):
        for y in range(height):
            rgb=palette[img.getpixel([x,y])]
            clst="#"+hex(rgb[0])[2:4]+hex(rgb[1])[2:4]+hex(rgb[2])[2:4]
            for z in range(3):
                if 0==rgb[z]:
                    clst=clst[:2*(z+1)]+"0"+clst[2*(z+1):]
            canvas.create_oval((10*x+1,10*y+1),(10*(x+1),10*(y+1)),fill=clst)

WriteAll()
root.mainloop()
```
